```python
import pandas as pd
import numpy as np

print("Pandas and NumPy imported successfully!")
```

```python
data = {
    'Order_ID': [101, 102, 103, 104, 105],
    'Product': ['Laptop', 'Mobile', 'Mouse', 'Keyboard', 'Laptop'],
    'Quantity': [2, 3, 5, 4, 1],
    'Price': [50000, 15000, 500, 1000, 50000],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai'],
    'Payment': ['UPI', 'Cash', 'Card', 'UPI', 'Card']
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Total_Sales'] = df['Quantity'] * df['Price']

print(df)
```

```python
total_revenue = np.sum(df['Total_Sales'])

print("Total Revenue:", total_revenue)
```

```python
print(df['Total_Sales'])
print("Total Revenue:", np.sum(df['Total_Sales']))
```

```python
average_sales = np.mean(df['Total_Sales'])

print("Average Sales:", average_sales)
```

```python
highest_sales = np.max(df['Total_Sales'])

print("Highest Sales:", highest_sales)
```

```python
lowest_sales = np.min(df['Total_Sales'])

print("Lowest Sales:", lowest_sales)
```

```python
product_sales = df.groupby('Product')['Total_Sales'].sum()

print(product_sales)
```

```python
city_sales = df.groupby('City')['Total_Sales'].sum()

print(city_sales)
```

```python
payment_sales = df.groupby('Payment')['Total_Sales'].sum()

print(payment_sales)
```

```python
total_orders = df['Order_ID'].count()

print("Total Orders:", total_orders)
```

```python
print(df['Total_Sales'].describe())
```

```python
Q1 = np.percentile(df['Total_Sales'], 25)
Q3 = np.percentile(df['Total_Sales'], 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['Total_Sales'] < lower_limit) |
              (df['Total_Sales'] > upper_limit)]

print("Outliers:")
print(outliers)
```

```python
print("===== RETAIL SALES PERFORMANCE REPORT =====")

print("Total Orders:", len(df))

print("Total Revenue:", df['Total_Sales'].sum())

print("Average Sales:", df['Total_Sales'].mean())

print("Highest Sales:", df['Total_Sales'].max())

print("Lowest Sales:", df['Total_Sales'].min())

print("Best Selling Product:")
print(product_sales.idxmax())

print("Best Performing City:")
print(city_sales.idxmax())

print("Most Used Payment Method:")
print(payment_sales.idxmax())
```

```python
import matplotlib.pyplot as plt

product_sales.plot(kind='bar')

plt.title("Product Wise Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Total Sales")

plt.show()
```

```python
city_sales.plot(kind='bar')

plt.title("City Wise Sales Analysis")
plt.xlabel("Cities")
plt.ylabel("Total Sales")

plt.show()
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 50

products = np.random.choice(
    ['Laptop', 'Mobile', 'Mouse', 'Keyboard', 'Monitor'], n
)

prices = {
    'Laptop': 50000,
    'Mobile': 15000,
    'Mouse': 500,
    'Keyboard': 1000,
    'Monitor': 12000
}

df = pd.DataFrame({
    'Order_ID': range(1001, 1051),
    'Order_Date': np.random.choice(
        pd.date_range('2026-01-01', '2026-06-30'), n
    ),
    'Customer_ID': np.random.choice(
        ['C101', 'C102', 'C103', 'C104', 'C105',
         'C106', 'C107', 'C108', 'C109', 'C110'], n
    ),
    'Product': products,
    'Quantity': np.random.randint(1, 6, n),
    'Price': [prices[p] for p in products],
    'City': np.random.choice(
        ['Delhi', 'Mumbai', 'Pune', 'Jaipur', 'Noida'], n
    ),
    'Payment': np.random.choice(
        ['UPI', 'Cash', 'Card'], n
    )
})

df['Total_Sales'] = df['Quantity'] * df['Price']

print(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

print("Data Types:")
print(df.dtypes)
```

```python
top_products = df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False).head(10)

print("Top 10 Products:")
print(top_products)
```

```python
city_revenue = df.groupby('City')['Total_Sales'].sum().sort_values(ascending=False)

print("City-wise Revenue Ranking:")
print(city_revenue)
```

```python
df['Month'] = df['Order_Date'].dt.to_period('M')

monthly_revenue = df.groupby('Month')['Total_Sales'].sum()

print("Monthly Revenue:")
print(monthly_revenue)
```

```python
top_customers = df.groupby('Customer_ID')['Total_Sales'].sum().sort_values(ascending=False).head(10)

print("Top 10 Customers:")
print(top_customers)
```

```python
payment_comparison = df.groupby('Payment')['Total_Sales'].sum().sort_values(ascending=False)

print("Payment Method Comparison:")
print(payment_comparison)
```

```python
print("Mean:", np.mean(df['Total_Sales']))

print("Median:", np.median(df['Total_Sales']))

print("Standard Deviation:", np.std(df['Total_Sales']))
```

```python
Q1 = np.percentile(df['Total_Sales'], 25)
Q3 = np.percentile(df['Total_Sales'], 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Total_Sales'] < lower_limit) |
    (df['Total_Sales'] > upper_limit)
]

print("Outliers:")
print(outliers)
```

```python
product_sales = df.groupby('Product')['Total_Sales'].sum()

product_sales.plot(kind='bar')

plt.title("Product Wise Sales Analysis")
plt.xlabel("Products")
plt.ylabel("Total Sales")

plt.show()
```

```python
city_revenue.plot(kind='bar')

plt.title("City Wise Revenue Analysis")
plt.xlabel("Cities")
plt.ylabel("Total Revenue")

plt.show()
```

```python
monthly_revenue.plot(kind='bar')

plt.title("Monthly Revenue Analysis")
plt.xlabel("Month")
plt.ylabel("Total Revenue")

plt.show()
```
